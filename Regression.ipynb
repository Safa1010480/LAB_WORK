```python
import numpy as np
```

```python
m = 9423.81532303
c = 24380.201479473726

inputs = [9.7, 5.0]

for x in inputs:
    y = m * x + c
    print(f"x = {x}  ->  y = {y:.4f}")
```

```python
coef = np.array([199.5132451, 10225.20442601, 8208.43477213,
                 993.53717098, 14885.38441545, 5146.14838675,
                 115.06859497])
c = -2010567.4201016556

X = np.array([
    [4012, 3, 1, 2016, 2.098092, 1, 5],
    [2310, 3, 1, 1988, 1.369622, 1, 4],
])

y = X @ coef + c          # same as np.dot(X, coef) + c
for row, pred in zip(X, y):
    print(row, "->", round(pred, 4))
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
coef = np.array([0.04559683, 0.09394991, -0.00964773,
                 0.00442035, 0.03010364, 0.48521205])
c = -9.2358725

X = np.array([
    [43, 34.0, 58.0, 190.0, 98,  0.430],
    [49, 30.4, 88.0, 169.5, 165, 0.302],
])

z = X @ coef + c
y_prob = sigmoid(z)
y_class = (y_prob >= 0.5).astype(int)   # threshold at 0.5

for row, zi, p, cl in zip(X, z, y_prob, y_class):
    print(f"{row} -> z = {zi:.4f}, probability = {p:.4f}, class = {cl}")
```
